## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My Lane as an ML Task

**Lane:** Refresh / Content Opportunity Scoring

**Task Type:** Ranking / Scoring

The goal of this project is to rank content pages according to their priority for review. Instead of simply classifying pages as "refresh" or "do not refresh," the model assigns a score that helps the SEO team decide which pages should be reviewed first. This makes the task a ranking/scoring problem because the output is an ordered list of pages based on priority.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

## Target or Proxy

The target variable is **is_declining_label**, which indicates whether a page is declining in performance.

This label is derived from observed page performance trends in the dataset. The model will learn patterns from features such as impressions, CTR, content age, average position, and word count to estimate the likelihood that a page should be reviewed for refresh.

The predicted scores will then be used to rank pages according to refresh priority.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

## Success Metric

The primary success metric for this project is **Precision@50**.

Precision@50 measures how many of the top 50 pages recommended by the model are actually declining.

This metric is appropriate because content teams usually review only a limited number of pages. Therefore, producing a high-quality ranking for the highest-priority pages is more valuable than maximizing overall accuracy.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create binary target
df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

print("Dataset Shape:", df.shape)

print("\nFirst Five Rows:")
display(df.head())

print("\nTarget Distribution:")
print(df["is_declining_label"].value_counts())
```

## Unit of Analysis

Each row in the dataset represents **one content page**.

For every page, the dataset contains observable features such as:

- Search volume
- Impressions
- CTR
- Average position
- Content age
- Days since last update
- Word count

The model makes one prediction for each content page, estimating its priority for refresh.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

## Why Machine Learning Beats a Fixed Rule

A fixed rule might rely on only one or two conditions, such as recommending pages that are old and have many impressions. However, content performance depends on many interacting factors, including impressions, CTR, average position, content age, and engagement.

Machine learning can learn these complex relationships automatically and combine multiple signals to produce a more accurate ranking than a manually designed rule.

The starter pipeline also showed that a learned model achieved a higher Precision@50 than the simple hand-written baseline, demonstrating that combining multiple features provides better decision support.
