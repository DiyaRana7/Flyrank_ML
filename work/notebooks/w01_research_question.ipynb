## Lane: Refresh / Content Opportunity Scoring

I chose the **Refresh / Content Opportunity Scoring** lane because the goal is to help prioritize which content pages should be reviewed first. Large websites often contain thousands of pages, making manual review difficult and time-consuming. Instead of checking every page individually, I want to build a data-driven ranking system that uses observable search and engagement signals to identify pages that are most likely to benefit from review. This project supports human decision-making by helping teams focus their efforts on the highest-priority pages.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

**Research Question:**
Which content pages should be prioritized for refresh based on observable search and engagement signals?

### Decision
The decision is determining which content pages should be reviewed first for refresh, allowing the SEO team to prioritize limited time and resources effectively.

### Action
The SEO or content team can review the highest-ranked pages and decide whether to:
- update outdated content
- improve metadata
- expand information
- improve page structure
- monitor the page

### Cost of a Wrong Recommendation

**False Positive**
A page is recommended for refresh even though it does not need changes. This wastes time and resources.

**False Negative**
A page that actually needs refresh is missed. As a result, valuable content may continue losing traffic and visibility.

### Why Machine Learning?

Machine learning can analyze multiple signals such as impressions, CTR, content age, average position, and engagement together. Instead of manually reviewing thousands of pages, it helps rank pages based on their likelihood of needing review, making the decision process faster and more efficient.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total Pages:", len(df))

print("\nCorrelation between Search Volume and Impressions")
print(df["search_volume"].corr(df["impressions_90d"]))

print("\nMedian Word Count by Trend")
print(df.groupby("trend_direction")["word_count"].median())

print("\nTrend Direction Distribution")
print(df["trend_direction"].value_counts())
```

### Observations

- The starter dataset contains **30,000 pages**, providing enough examples for exploratory analysis.
- The correlation between **search volume** and **impressions** is **0.0012**, showing that keyword search volume alone is not a strong predictor of page traffic.
- The median word count for **declining pages (2909 words)** and **growing pages (2847.5 words)** is very similar, suggesting that content length alone is unlikely to explain performance differences.
- The dataset contains **16,262 declining pages**, indicating that identifying refresh opportunities is a meaningful and practical problem for this project.

These observations suggest that multiple search and engagement signals should be considered together when deciding which pages deserve review.
